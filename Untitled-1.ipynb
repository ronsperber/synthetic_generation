```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# ----------------------------
# 1. Load and normalize iris
# ----------------------------
from sklearn.datasets import load_iris
iris = load_iris()
X = iris.data.astype(np.float32)
y = iris.target.reshape(-1,1)

scaler = StandardScaler()
X_norm = scaler.fit_transform(X)

# One-hot encode species
enc = OneHotEncoder(sparse_output=False)
C = enc.fit_transform(y).astype(np.float32)

x_real = torch.tensor(X_norm)
c_real = torch.tensor(C)

noise_dim = 4  # same as feature dimension for simplicity

# ----------------------------
# 2. Conditional GAN networks
# ----------------------------
class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(noise_dim, 32),
            nn.BatchNorm1d(32),
            nn.LeakyReLU(0.2),
            nn.Linear(32, 32),
            nn.BatchNorm1d(32),
            nn.LeakyReLU(0.2),
            nn.Linear(32, 4)
        )
    def forward(self, z):
        return self.net(z)

class Discriminator(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(4, 32),
            nn.LeakyReLU(0.2),
            nn.Linear(32,32),
            nn.LeakyReLU(0.2),
            nn.Linear(32,1),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x)

G = Generator()
D = Discriminator()
opt_G = optim.Adam(G.parameters(), lr=0.001)
opt_D = optim.Adam(D.parameters(), lr=0.001)

# ----------------------------
# 3. Training loop
# ----------------------------
num_epochs = 5000
batch_size = 32

for epoch in range(num_epochs):
    idx = torch.randint(0, x_real.size(0), (batch_size,))
    real_batch = x_real[idx]
    
    
    # --- Discriminator ---
    z = torch.randn(batch_size, noise_dim)
    fake_batch = G(z).detach()
    
    D_real = D(real_batch)
    D_fake = D(fake_batch)
    
    loss_D = -torch.mean(torch.log(D_real + 1e-8) + torch.log(1 - D_fake + 1e-8))
    
    opt_D.zero_grad()
    loss_D.backward()
    opt_D.step()
    
    # --- Generator ---
    z = torch.randn(batch_size, noise_dim)
    fake_batch = G(z)
    D_fake = D(fake_batch)
    
    loss_G = -torch.mean(torch.log(D_fake + 1e-8))
    
    opt_G.zero_grad()
    loss_G.backward()
    opt_G.step()

# ----------------------------
# 4. Generate samples
# ----------------------------
def generate_samples(n=150):
    z = torch.randn(n, noise_dim)
    
    samples = G(z).detach().numpy()
    return samples

samples_norm = generate_samples()
# Un-normalize to original scale
samples = scaler.inverse_transform(samples_norm)

```

```python
class Critic(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(4, 32),
            nn.LeakyReLU(0.2),
            nn.Linear(32, 32),
            nn.LeakyReLU(0.2),
            nn.Linear(32, 1)
        )

    def forward(self, x):
        return self.net(x)
```

```python
G = Generator()
D = Critic()

opt_G = optim.Adam(G.parameters(), lr=1e-4)
opt_D = optim.Adam(D.parameters(), lr=1e-4)

clip_value = 0.01
n_critic = 5  # train critic more than generator

for epoch in range(5000):
    for _ in range(n_critic):
        idx = torch.randint(0, x_real.size(0), (32,))
        real = x_real[idx]

        z = torch.randn(32, noise_dim)
        fake = G(z).detach()

        loss_D = -(D(real).mean() - D(fake).mean())

        opt_D.zero_grad()
        loss_D.backward()
        opt_D.step()

        # Weight clipping
        for p in D.parameters():
            p.data.clamp_(-clip_value, clip_value)

    # --- Generator ---
    z = torch.randn(32, noise_dim)
    fake = G(z)
    loss_G = -D(fake).mean()

    opt_G.zero_grad()
    loss_G.backward()
    opt_G.step()
```
