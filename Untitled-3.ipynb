```python
import numpy as np

def make_gaussian_mixture(
    n_samples=50_000,
    centers=8,
    radius=5.0,
    std=0.5,
    seed=0
):
    rng = np.random.default_rng(seed)

    angles = np.linspace(0, 2*np.pi, centers, endpoint=False)
    means = np.stack([
        radius * np.cos(angles),
        radius * np.sin(angles)
    ], axis=1)

    samples = []
    for _ in range(n_samples):
        k = rng.integers(0, centers)
        x = rng.normal(loc=means[k], scale=std, size=2)
        samples.append(x)

    return np.array(samples, dtype=np.float32)
```

```python
import torch
import torch.nn as nn

class Generator(nn.Module):
    def __init__(self, z_dim=2, hidden=128):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(z_dim, hidden),
            nn.ReLU(),
            nn.Linear(hidden, hidden),
            nn.ReLU(),
            nn.Linear(hidden, 2),
        )

    def forward(self, z):
        return self.net(z)


class Discriminator(nn.Module):
    def __init__(self, hidden=128):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, hidden),
            nn.LeakyReLU(0.2),
            nn.Linear(hidden, hidden),
            nn.LeakyReLU(0.2),
            nn.Linear(hidden, 1),
            nn.Sigmoid(),
        )

    def forward(self, x):
        return self.net(x)
```

```python
import matplotlib.pyplot as plt

@torch.no_grad()
def plot_samples(G, X_real, n=2000):
    z = torch.randn(n, 2, device=next(G.parameters()).device)
    X_fake = G(z).cpu().numpy()

    plt.figure(figsize=(5,5))
    plt.scatter(X_real[:n,0], X_real[:n,1], s=5, alpha=0.3, label="Real")
    plt.scatter(X_fake[:,0], X_fake[:,1], s=5, alpha=0.3, label="Fake")
    plt.legend()
    plt.axis("equal")
    plt.show()
```

```python
from torch.utils.data import DataLoader, TensorDataset

# Data
X = make_gaussian_mixture()
dataset = TensorDataset(torch.tensor(X))
loader = DataLoader(dataset, batch_size=512, shuffle=True)

# Models
G = Generator(z_dim=2)
D = Discriminator()

# Optimizers
g_opt = torch.optim.Adam(G.parameters(), lr=1e-4, betas=(0.5, 0.9))
d_opt = torch.optim.Adam(D.parameters(), lr=1e-4, betas=(0.5, 0.9))

criterion = nn.BCELoss()

device = "cuda" if torch.cuda.is_available() else "cpu"
G.to(device)
D.to(device)

for epoch in range(1,501):
    for (x_real,) in loader:
        x_real = x_real.to(device)

        # ---------------------
        # Train Discriminator
        # ---------------------
        z = torch.randn(len(x_real), 2, device=device)
        x_fake = G(z).detach()

        d_real = D(x_real)
        d_fake = D(x_fake)

        loss_d = (
            criterion(d_real, torch.ones_like(d_real)) +
            criterion(d_fake, torch.zeros_like(d_fake))
        )

        d_opt.zero_grad()
        loss_d.backward()
        d_opt.step()

        # ---------------------
        # Train Generator
        # ---------------------
        z = torch.randn(len(x_real), 2, device=device)
        x_fake = G(z)
        d_fake = D(x_fake)

        loss_g = criterion(d_fake, torch.ones_like(d_fake))

        g_opt.zero_grad()
        loss_g.backward()
        g_opt.step()
    if epoch % 50 == 0:
        plot_samples(G, x_real)

    if epoch % 20 == 0:
        print(
            f"Epoch {epoch:03d} | "
            f"D(real)={d_real.mean().item():.3f} "
            f"D(fake)={d_fake.mean().item():.3f}"
        )
```

```python
plot_samples(G, x_real, n=x_real.shape[0])
```
